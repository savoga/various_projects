```python
import numpy as np

import matplotlib.pyplot as plt
```

## Normal

```python
data = np.random.normal(0,1,10000)
plt.hist(data)
plt.title('Normal distribution (mu=0, sigma=1, n=10000)')
plt.show()
```

## Binomial

```python
data = np.random.binomial(100,0.2, size=10000)
plt.hist(data)
plt.title('Binomial distribution (n_trials=100, proba=0.2, size=10000)')
plt.show()
```

## Beta

```python
data = np.random.beta(4,6,size=10000)
plt.hist(data)
plt.title('Beta distribution (a=4, b=6, size=10000)')
plt.show()
```

## Pareto

```python
data = np.random.pareto(10, size=100)
plt.hist(data)
plt.title('Pareto distribution (a=2, size=100)')
plt.show()
```

## Logistic

```python
gamma1 = 1
gamma2 = 2

# Density
samples = np.random.logistic(loc=0, scale=1/gamma1, size=100_000)
plt.hist(samples, bins=50, density=True, alpha=.8, label="gamma={}".format(gamma1))

samples = np.random.logistic(loc=0, scale=1/gamma2, size=100_000)
plt.hist(samples, bins=50, density=True, alpha=.8, label="gamma={}".format(gamma2))

plt.title("Logistic distribution")
plt.legend()
plt.show

# CDF

plt.figure()
plt.title("Logistic (sigmoid) function")
plt.xlabel("z")
plt.ylabel("P")
plt.grid(True)

x = np.linspace(-10, 10, 500)
y = 1 / (1 + np.exp(-gamma1 * x))
plt.plot(x, y, label='gamma={}'.format(gamma1))

x = np.linspace(-10, 10, 500)
y = 1 / (1 + np.exp(-gamma2 * x))
plt.plot(x, y, label='gamma={}'.format(gamma2))

plt.legend()
plt.show()
```

Gamma low:

Density: values are more spread, heavier tails

CDF: transition phase is wider => proba goes **earlier** but **slower** to 1
